```python
import pandas as pd
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
```

## 1. Model Initialization

Initialize the Llama 3.2 model using Ollama.

```python
llm = ChatOllama(model="llama3.2")

response = llm.invoke("What is Artificial Intelligence?")
print(response.content)
```

## 2. Prompt Templates

Create five templates for General, Attendance, Exams, Assignments and Career.

```python
prompts = {
    "General": "You are a college assistant. Answer: {question}",
    "Attendance": "You are an attendance assistant. Answer: {question}",
    "Exams": "You are an exam assistant. Answer: {question}",
    "Assignments": "You are an assignment assistant. Answer: {question}",
    "Career": "You are a career assistant. Answer: {question}"
}

for category, template in prompts.items():
    print(category, ":", template)
```

## 3. Testing 20 User Queries

Test 20 questions using different prompt templates
and save the generated responses in a CSV file.

df = pd.read_csv("../dataset/user_queries.csv")

results = []

for _, row in df.iterrows():
    category = row["Category"]
    question = row["Question"]

    prompt = ChatPromptTemplate.from_template(
        prompts.get(category, prompts["General"])
    )

    chain = prompt | llm
    response = chain.invoke({"question": question})

    print("\nQuestion:", question)
    print("Response:", response.content)

    results.append({
        "Query_ID": row["Query_ID"],
        "Category": category,
        "Question": question,
        "Response": response.content
    })

pd.DataFrame(results).to_csv(
    "../dataset/chatbot_responses.csv", index=False
)

print("All 20 queries tested successfully!")

## 4. Prompt Comparison

Compare basic and improved prompts using the first five user queries.
The responses are generated using the Llama 3.2 model.

```python
df = pd.read_csv("../dataset/user_queries.csv")

basic_prompt = ChatPromptTemplate.from_template(
    "Answer this question in simple English: {question}"
)

improved_prompt = ChatPromptTemplate.from_template(
    "You are a helpful college assistant. "
    "Give a clear and detailed answer: {question}"
)

results = []

for _, row in df.head(5).iterrows():
    question = row["Question"]

    basic = (basic_prompt | llm).invoke(
        {"question": question}
    )

    improved = (improved_prompt | llm).invoke(
        {"question": question}
    )

    print("\nQuestion:", question)
    print("\nBasic Response:", basic.content)
    print("\nImproved Response:", improved.content)

    results.append({
        "Question": question,
        "Basic_Response": basic.content,
        "Improved_Response": improved.content
    })

pd.DataFrame(results).to_csv(
    "../dataset/prompt_comparison.csv", index=False
)

print("\nPrompt comparison completed successfully!")
```

## 5. Summarization Chain

Summarize a given paragraph using LangChain and the Llama 3.2 model.
The summary is generated in exactly five simple points.

```python
text = """
College education helps students develop knowledge,
communication skills, critical thinking and confidence.
It provides opportunities to participate in activities,
work with others and prepare for future careers.
Students also learn time management, problem solving
and responsibility through their college experiences.
"""

summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following text in exactly 5 simple points:\n{text}"
)

summary_chain = summary_prompt | llm

summary = summary_chain.invoke({"text": text})

print("Original Text:\n", text)
print("\nSummary:\n", summary.content)
```

## 6. Question Answering Chain

Use LangChain to answer questions based on the given context.
The Llama 3.2 model generates answers using the provided information.

```python
context = """
College students develop communication skills, teamwork,
problem-solving, time management, critical thinking,
leadership, technical skills, creativity and confidence.
These skills help students prepare for their future careers.
"""

qa_prompt = ChatPromptTemplate.from_template(
    "Answer the question using the context below.\n"
    "Give the answer in simple English in about 3 lines.\n"
    "Context: {context}\n"
    "Question: {question}"
)

qa_chain = qa_prompt | llm

question = "What skills do college students develop?"

answer = qa_chain.invoke({
    "context": context,
    "question": question
})

print("Question:", question)
print("\nAnswer:", answer.content)
```
